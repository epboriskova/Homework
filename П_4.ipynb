```python
#Считайте датасет из файла train.csv (это данные о выживаемости на Титанике)
import pandas as pd
df = pd.read_csv('/content/gdrive/MyDrive/train.csv')
```

```python
import matplotlib.pyplot as plt
```

```python
#Визуализируйте распределение значений признаков Survived, Pclass, Age, Sex. Parch
x = df['Survived'].unique()
y = df['Survived'].value_counts()
fig, ax = plt.subplots()
ax.bar(x=x, height=y)
plt.title('Распределение значений признака Survived')
plt.xlabel('Выжившие/невыжившие', color='red')
plt.ylabel('Количество', color='green')
```

```python
x = df['Sex'].unique()
y = df['Sex'].value_counts()
fig, ax = plt.subplots()
ax.bar(x=x, height=y)
plt.title('Распределение значений признака Sex')
plt.xlabel('Sex', color='red')
plt.ylabel('Количество', color='green')
```

```python
x = df['Parch'].unique()
y = df['Parch'].value_counts()
fig, ax = plt.subplots()
ax.bar(x=x, height=y)
plt.title('Распределение значений признака Parch')
plt.xlabel('Parch', color='red')
plt.ylabel('Количество', color='green')
```

```python
#Постройте график типа boxplot для столбца Age
import seaborn as sns
sns.set(rc={'figure.figsize':(10, 8)})
sns.boxplot(y = df['Age'])
plt.title('График типа boxplot для столбца Age')
```

#Проинтепретируйте полученный график
По графику видно, что большая часть пассажиров относится к возрастной группе от 20 до 40 лет. Пассажиров старше 65 лет на борту практически не было. Самому пожилому пассажиру было 80 лет, а самый молодой был практически новорожденным. Средний возраст пассажиров был менее 30 лет.

```python
#Постройте график типа pie chart для переменных Survived, Pclass, подпишите доли в процентах
pie_1 = df.groupby('Pclass')['Survived'].value_counts()
fig, ax = plt.subplots()
ax.pie(pie_1, autopct= ' %1.1f%%')
plt.title('Соотношение выживших и невыживших пассажиров, а также их распредение по классам')
```

```python
import plotly.express as px
```

```python
#Постройте интерактивный sunburst plot (визуализация иерархических данных) с помощью plotly. На первом уровне иерархии - количество пассажиров в каждом из классов, а на втором количество женщин/мужчин в этом классе
value_1 = df['Pclass'].value_counts().tolist()
value_1.insert(0, len(df))
value_2 = df.groupby('Sex')['Pclass'].value_counts().tolist()
value_3 = value_1 + value_2
print(value_1)
print(value_2)
print(value_3)

data = dict(
    categories = ['Total', '3', '1', '2', 'F_3', 'F_1', 'F_2', 'M_3', 'M_1', 'M_2'],
    parent = ['', 'Total', 'Total', 'Total', '3', '1', '2', '3', '1', '2'],
    value = value_3
)

fig = px.sunburst(data, names='categories', values = 'value', parents='parent', branchvalues='total')
fig.show()
```
